## Kiến trúc



Phần ảnh giữ đúng p16. Hai chỗ gắn nhãn **mới** là phần run này thêm cho IMU: **(B) CNN làm mượt
IMU** sau Decoder IMU, và **(C) rung thừa** trong Loss phase 2. Hình: `docs/kien_truc_imu.svg`, vẽ lại
bằng `python3 tools/draw_architecture_imu.py docs/kien_truc_imu.svg`.

1. **① Backbone (xanh dương).** Ảnh mờ + tối (nhiễu B) qua **QWT Hilbert**, IMU nhiễu qua
   **Haar**; mỗi bên một encoder CNN 4 stage; fusion có cổng trộn hai bên thành latent **ZI**
   (ảnh, 128 × 16 × 16) và **ZU** (IMU, 128 × 8). Train ở phase 1, đóng băng ở phase 2.
2. **② Phase 1 (vàng, chỉ lúc train).** Teacher EMA (bản sao hai encoder, m tới 0,9995) đọc bản
   **sạch**, cho ra đích **TI** (16×16, kèm đích **mịn** 32×32) và **TU**. Hai predictor riêng,
   **ZI → TI** và **ZU → TU**, **nhìn lân cận** (ảnh 5×5 token); 30% token ảnh và 25% token IMU bị
   **che** và phải đoán từ lân cận. Loss JEPA (+ đích mịn, đích thô 8×8), VICReg với covariance
   **gộp**, decoder neo và **Jacobian** train backbone; Jacobian ép encoder nhạy với đường nét,
   bỏ qua nhiễu.
3. **③ Phase 2 (xanh lá).** Backbone **đóng băng**. Decoder ảnh nhận ZI và **chính ảnh mờ** (skip),
   gồm hai nhánh rồi **ghép**:
   - Nhánh **màu** ở 128×128: một đường cong tone + ma trận màu **cho cả ảnh** (gỡ gamma, cân bằng
     trắng), rồi ResNet sửa màu từng vùng.
   - Nhánh **đường nét** (kênh sáng Y, 256×256) là **NAFNet** — U-Net 5 tầng, 30 NAFBlock
     (LayerNorm, conv depthwise, SimpleGate, channel attention). **Encoder** thu nhỏ ÷2 mỗi tầng,
     từ **đặc trưng nhỏ** (cạnh, vật nhỏ; tầng 256² có 48 kênh × 3 + 3 khối) tới **đặc trưng
     tổng thể** (bố cục, vật là gì); **Bottleneck** 16×16 nhận **ZI** (8 khối); **Decoder** phóng
     ×2 trở lại; **skip connection** cộng đặc trưng nhỏ từ Encoder sang Decoder; đầu phụ ở 128² và
     64² được chấm riêng.
   - **CNN làm nét + mượt** (từ p15) chạy sau NAFNet, chỉ trên bản đồ chi tiết Y — màu đã tách
     riêng nên nó không chạm được màu: 4 khối residual × 32 kênh ở 256², không thu nhỏ.

   - **IMU (mới ở run này):** decoder IMU (hệ số Haar, skip có cổng) nhận ZU, rồi **(B) CNN làm mượt
     IMU**: CNN 1-D, 4 khối × 32 kênh, dilation 1-2-4-8 nên nhìn 65 mẫu (0,65 s), đọc cả IMU đã khôi
     phục lẫn IMU nhiễu, lúc đầu trả nguyên đầu ra decoder (27 K tham số).

   Decoder IMU và CNN làm mượt nhận ZU. **Loss phase 2** (hồng, góc phải dưới) chỉ train
   hai decoder: ảnh — L1, chi tiết QWT, VGG16 perceptual; màu — L1 + thống kê màu; nét — L1, độ dốc,
   FFT phức, nhiều tỉ lệ, gồ ghề thừa (ép mượt); IMU — L1, chi tiết Haar, độ rung, và **(C) rung thừa** (trọng số 2,0: chỉ phạt thay đổi giữa hai mẫu
   **vượt** tín hiệu sạch). Từ p16, loss
   VGG có trọng số 0,5 và tính trên mảnh cắt 128×128.

**p19 thêm vào hình trên:** (e) mũi tên từ predictor ZI → TI của phase 1 (giữ lại, đóng băng) vào Bottleneck NAFNet cạnh
ZI; (f) backbone hết đóng băng ở 25% update cuối; với `p19_sharp`: (b) encoder không GroupNorm, (c) đầu "hư hỏng" từ ZI.

## Cell 3 — Xác nhận nhanh dataset root và vài timeline; Cell 5 kiểm tra toàn bộ


```python
cell_started = time.perf_counter()
from collections import Counter
from qjepa.data.tartanair import Trajectory, discover_trajectories, audit_trajectory

if TARTANAIR640:
    from qjepa.data.tartanair import find_shard_roots, link_trajectories
    # Mỗi shard là một dataset, quỹ đạo nằm ở độ sâu khác nhau. Thiếu shard nào trong SHARDS thì dừng.
    shard_roots = find_shard_roots(INPUT, shards=SHARDS)
    for shard, root in shard_roots.items():
        print('shard', shard, root)
    DATA_ROOT = link_trajectories(list(shard_roots.values()), Path('/tmp/tartanair640_all'))


def candidate_roots(base):
    # Không duyệt hàng chục nghìn file ảnh khi dò thư mục IMU.
    found = Counter()
    for directory, children, files in os.walk(base):
        children[:] = [name for name in children if name != 'image_lcam_front']
        if Path(directory).name != 'imu' or not {'imu_time.npy', 'imu_time.txt'} & set(files):
            continue
        trajectory = Path(directory).parent
        if (trajectory / 'image_lcam_front').is_dir() and len(trajectory.parents) >= 3:
            found[trajectory.parents[2]] += 1
    return found

if DATA_ROOT is None:
    counts = Counter()
    for mount in sorted(INPUT.iterdir()):
        if mount.is_dir():
            counts.update(candidate_roots(mount))
    assert counts, 'Không thấy trajectory nào; kiểm tra dataset đã gắn và đã giải nén.'
    for root, total in counts.most_common():
        print(f'{total:5d} trajectory  {root}')
    DATA_ROOT = counts.most_common(1)[0][0]
    if len(counts) > 1:
        print('Nếu root được chọn sai, đặt DATA_ROOT ở Cell 1 rồi chạy lại.')

DATA_ROOT = Path(DATA_ROOT)
assert DATA_ROOT.is_dir(), f'Thiếu dataset root: {DATA_ROOT}'
# Glob đúng ba tầng env/difficulty/trajectory, không mở các thư mục ảnh.
paths = {}
for pattern in ('*/*/*/imu/imu_time.npy', '*/*/*/imu/imu_time.txt'):
    for timeline in DATA_ROOT.glob(pattern):
        paths.setdefault(timeline.parent.parent, timeline)
if paths:
    trajectories = []
    for path in sorted(paths):
        assert (path / 'image_lcam_front').is_dir(), f'Thiếu ảnh: {path}'
        environment, difficulty, trajectory_id = path.relative_to(DATA_ROOT).parts
        trajectories.append(Trajectory(path, environment, difficulty, trajectory_id))
else:
    # Chỉ dùng đường quét tổng quát khi dataset có split/subfolder khác layout chuẩn.
    trajectories = discover_trajectories(DATA_ROOT)
keys = [trajectory.key for trajectory in trajectories]
assert keys and len(keys) == len(set(keys)), 'Không tìm thấy trajectory hoặc DATA_ROOT đang gộp bản sao.'
hints = {trajectory.split_hint for trajectory in trajectories}
assert len(hints) == 1, f'Split hint không đồng nhất {hints}; DATA_ROOT đang trỏ quá cao.'
print('DATA_ROOT:', DATA_ROOT, '| trajectory:', len(trajectories),
      '| motion key:', len({trajectory.motion_key for trajectory in trajectories}),
      '| split hint:', next(iter(hints)))
print('Difficulty:', dict(Counter(trajectory.difficulty for trajectory in trajectories)))
for trajectory in trajectories[:3]:
    print(audit_trajectory(trajectory, window=128))
del trajectories
gc.collect()
print('Cell 3 giây:', round(time.perf_counter() - cell_started, 1))
```

```python
cell_started = time.perf_counter()
from qjepa.config import load_config, serializable_config, validate_config
from qjepa.training.checkpoints import configuration_hash, load_checkpoint, require_phase1_checkpoint

CONFIG_NAME = {'p19_sharp_p2': 'kaggle_sharp_p2.yaml', 'p19_sharp': 'kaggle_sharp.yaml'}[RUN]
OUT = PROJECT / 'outputs' / RUN
MANIFEST = PROJECT / 'manifests/kaggle'
CONFIG = PROJECT / 'configs/kaggle.yaml'
P1_UPDATES = 5000  # bang p5 (phase 1 cua p8) de so sanh co kiem soat
P2_UPDATES = 20000  # recipe p19; LP-FT tu 75% so update
# p19_sharp_p2 chi doi phase 2: dung lai phase 1 cua mot run truoc ('auto' = do theo REUSE_RUN_NAMES: truoc het
# /kaggle/working, roi cac Input toi da 3 tang). Phase 1 dung lai mang theo recipe cua no; khong thay thi train moi
# theo recipe p16_infomax. p19_sharp doi phase 1 nen luon train moi (None).
REUSE_PHASE1_FROM = 'auto' if RUN == 'p19_sharp_p2' else None
REUSE_RUN_NAMES = ('p16_infomax', 'p16_imu_smooth_640', 'p16_fourier_imu', 'p16_imu_smooth', 'p16_1gpu', 'p16_amp_vgg')
REUSE_REQUESTED = REUSE_PHASE1_FROM is not None

if REUSE_PHASE1_FROM == 'auto':
    candidates = [PROJECT / 'outputs' / name for name in REUSE_RUN_NAMES]
    for name in REUSE_RUN_NAMES:
        for depth in ('*', '*/*', '*/*/*'):
            candidates += sorted(p.parent.parent for p in
                                 INPUT.glob(f'{depth}/outputs/{name}/phase1/last.pt'))
    REUSE_PHASE1_FROM = next((c for c in candidates if (c / 'phase1/last.pt').is_file()), None)
    if REUSE_PHASE1_FROM is None and RESUME_RUN is None and not (OUT / 'phase1/last.pt').exists():
        print(f'!!! KHONG tim thay phase 1 cua {" / ".join(REUSE_RUN_NAMES)} (da tim /kaggle/working va /kaggle/input).')
        print('!!! Cell 8 se TRAIN PHASE 1 MOI theo recipe trong repo (5000 update, them vai gio GPU),')
        print(f'!!! va ket qua KHONG con so thang voi {REUSE_RUN_NAMES[0]}. Muon so thang: dung lai, gan archive')
        print(f'!!! {REUSE_RUN_NAMES[0]} lam Input roi chay lai Cell 4.')

# Recipe nam trong configs/kaggle_tartanair_v2.yaml. Cell nay chi ghi de nhung
# thu thuoc ve PHIEN CHAY (so update, duong dan, thiet bi) — khong ghi de cong
# thuc. Ban cu hardcode ca encoder_skips/imu_variation_weight o day, nen sua file
# config khong co tac dung va mot run da chay sai het: encoder_skips ve False du
# config ghi true, va imu_variation_weight ket o 0.5 du config ghi 2.0.
config = serializable_config(load_config(PROJECT / 'configs' / CONFIG_NAME))
config['phase1']['max_successful_updates'] = P1_UPDATES
config['phase2']['max_successful_updates'] = P2_UPDATES
if P2_UPDATES != 20000:
    # Recipe: backbone train cung tu update 15000/20000. Doi so update thi giu ty le 75%.
    config['phase2']['backbone_finetune_after_updates'] = int(0.75 * P2_UPDATES)
if RESUME_RUN is not None:
    saved = RESUME_RUN / 'config.yaml'
    assert saved.is_file(), 'Resume cần config.yaml đầy đủ trong archive của notebook này.'
    config = yaml.safe_load(saved.read_text())
    for phase in ('phase1', 'phase2'):
        resolved = RESUME_RUN / phase / 'resolved_config.yaml'
        if resolved.is_file():
            previous = yaml.safe_load(resolved.read_text())
            assert configuration_hash(previous, phase) == configuration_hash(config, phase), (
                f'Archive chứa config không khớp {phase}.'
            )
    if not OUT.exists():
        # Resume chỉ cần checkpoint, config và log; không sao chép các ảnh/eval cũ.
        keep = ('config.yaml', 'source_revision.txt',
                'phase1/last.pt', 'phase1/train.jsonl', 'phase1/resolved_config.yaml',
                'phase2/last.pt', 'phase2/best_joint_validation.pt',
                'phase2/best_blur_validation.pt', 'phase2/best_guarded_validation.pt',
                'phase2/train.jsonl', 'phase2/resolved_config.yaml',
                'phase2/guard_reference.json')
        restored = []
        for name in keep:
            source = RESUME_RUN / name
            if source.is_file():
                destination = OUT / name
                destination.parent.mkdir(parents=True, exist_ok=True)
                shutil.copy2(source, destination)
                restored.append(name)
        print('Đã khôi phục file cần để resume:', restored)
    else:
        print('OUT đã tồn tại: dùng bản local; không ghi đè từ archive.')

if REUSE_PHASE1_FROM is not None and RESUME_RUN is None and not (OUT / 'phase1/last.pt').exists():
    source = Path(REUSE_PHASE1_FROM) / 'phase1'
    assert (source / 'last.pt').is_file(), (
        f'Khong thay {source}/last.pt. Dat REUSE_PHASE1_FROM = None de train phase 1 moi, '
        "hoac 'auto' de tu do archive p8/p5.")
    # Chep ca thu muc: Cell 9 va Cell 11 doc them log, bank validation va duong cong.
    shutil.copytree(source, OUT / 'phase1', dirs_exist_ok=True)
    print('Dung lai phase 1 tu', source)
    print('  cac kiem tra hash ben duoi va o Cell 8 van chay tren ban chep —',
          'phase 1 khong khop se bi tu choi, khong am tham dung sai.')

if REUSE_REQUESTED and RESUME_RUN is None and (OUT / 'phase1/resolved_config.yaml').is_file():
    # Phase 1 dung lai mang theo recipe cua no: lay nguyen MOI muc chi phase 1 doc (phase1, encoder_sensitivity,
    # monitor). p16_infomax them san Jacobian vao encoder_sensitivity, nen chi lay muc phase1 la hash lech.
    # Cac muc phase 2 cung doc (model, data, corruption) van phai khop: kiem tra hash ben duoi dung lai neu khong.
    previous = yaml.safe_load((OUT / 'phase1/resolved_config.yaml').read_text())
    for section in ('phase1', 'encoder_sensitivity', 'monitor'):
        if previous[section] != config[section]:
            config[section] = previous[section]
            print(f'Muc {section} lay tu run duoc dung lai (khong phai recipe trong repo).')

config['data'].update(root=str(DATA_ROOT), manifest_dir=str(MANIFEST),
                      num_workers=DATALOADER_WORKERS, pin_memory=PIN_MEMORY)
if TARTANAIR640:
    config['data']['split_rule'] = 'per_environment'  # valid/test lấy từ mọi môi trường, không từ cả bể
if RESUME_RUN is None:
    # Recipe ghi amp_fp16; fp16 tung cham 5x tren T4 (p16), nen mac dinh fp32 nhu moi run p16. Doi o Cell 1.
    config['phase2']['precision'] = PRECISION
config['runtime'].update(output_dir=str(OUT), device='cuda', gpu_count=GPU_COUNT, parallel=PARALLEL,
                         cudnn_enabled=CUDNN)
validate_config(config)
for phase in ('phase1', 'phase2'):
    resolved = OUT / phase / 'resolved_config.yaml'
    if resolved.is_file():
        previous = yaml.safe_load(resolved.read_text())
        assert configuration_hash(previous, phase) == configuration_hash(config, phase), (
            f'{phase}: cấu hình khác run đang có. Chọn OUT mới hoặc dùng đúng config cũ.'
        )
OUT.mkdir(parents=True, exist_ok=True)
revision = OUT / 'source_revision.txt'
if revision.exists():
    assert revision.read_text().strip() == SOURCE_COMMIT, 'OUT thuộc source khác.'
revision.write_text(SOURCE_COMMIT + '\n', encoding='utf-8')
encoded = yaml.safe_dump(config, sort_keys=False)
CONFIG.write_text(encoded, encoding='utf-8')
(OUT / 'config.yaml').write_text(encoded, encoding='utf-8')
PHASE1 = OUT / 'phase1/last.pt'
PHASE2 = OUT / 'phase2/last.pt'
BEST = OUT / 'phase2/best_joint_validation.pt'
for phase, fields in {
    'phase1': ('max_successful_updates', 'coefficient_reconstruction_loss_weight', 'reconstruction_detail_weight',
               'covariance_pooling', 'predictor_type', 'image_mask_ratio', 'multiscale_fine_weight',
               'degradation_weight', 'predictor_degradation_condition', 'augment_hflip'),
    'phase2': ('max_successful_updates', 'encoder_skips', 'residual_sees_input', 'smooth_l1_beta',
               'reconstruction_detail_weight', 'imu_variation_weight', 'detail_energy_weight',
               'image_detail_loss', 'image_detail_source', 'image_decoder',
               'split_branch_arch', 'split_edge_naf_widths', 'split_edge_fft_weight',
               'split_edge_aux_weight', 'perceptual_weight', 'perceptual_crop', 'precision',
               'split_edge_refiner_blocks', 'split_edge_refiner_width', 'split_edge_refiner_scale',
               'split_edge_naf_enc_blocks', 'split_edge_naf_dec_blocks',
               'imu_refiner_blocks', 'imu_refiner_width', 'imu_jitter_weight',
               'split_edge_smooth_weight', 'split_gradient_weight',
               'split_color_global', 'split_color_stats_weight',
               'augment_hflip', 'decoder_predictor_input', 'backbone_finetune_after_updates',
               'backbone_finetune_lr_scale', 'imu_increment_weight', 'imu_increment_windows'),
}.items():
    print(phase, {key: config[phase].get(key, '—') for key in fields})
print('hash phase 1:', configuration_hash(config, 'phase1')[:8], '| phase 2:',
      configuration_hash(config, 'phase2')[:8], f'({RUN}; ghi lai de doi chieu lan sau)')
print('encoder_norm:', config['model'].get('encoder_norm', 'group'), '| Jacobian noise_direction:',
      config['encoder_sensitivity'].get('noise_direction', 'corruption'))
print('nhieu anh:', {key: config['corruption']['image'].get(key, '—') for key in (
    'defocus_sigma_px', 'motion_length_px', 'photon_count', 'noise_only_probability',
    'low_light_only_probability')})
print('OUT:', OUT)
print('workers:', config['data']['num_workers'], '| pin_memory:', config['data']['pin_memory'],
      '| GPU phase 2:', GPU_COUNT, PARALLEL, '| cuDNN:', CUDNN)

def run_cli(*arguments):
    command = [sys.executable, '-u', '-m', 'qjepa', *map(str, arguments)]
    print('Running:', ' '.join(command), flush=True)
    subprocess.run(command, cwd=PROJECT, check=True)

def run_tool(script, output, *arguments):
    command = [sys.executable, '-u', str(script), *map(str, arguments)]
    output = Path(output)
    output.parent.mkdir(parents=True, exist_ok=True)
    print('Running:', ' '.join(command), flush=True)
    with output.open('w', encoding='utf-8') as log:
        process = subprocess.Popen(command, cwd=PROJECT, stdout=subprocess.PIPE,
                                   stderr=subprocess.STDOUT, text=True, bufsize=1)
        try:
            for line in process.stdout:
                print(line, end='', flush=True)
                log.write(line)
            result = process.wait()
        except BaseException:
            process.terminate()
            try:
                process.wait(timeout=10)
            except subprocess.TimeoutExpired:
                process.kill()
                process.wait()
            raise
        finally:
            process.stdout.close()
        if result:
            raise subprocess.CalledProcessError(result, command)
print('Cell 4 giây:', round(time.perf_counter() - cell_started, 1))
```

```python
import zipfile

def export_run():
    assert PHASE1.is_file(), 'Chưa có checkpoint phase 1 để lưu.'
    export_dir = Path('/kaggle/working')
    stage = 'phase2' if PHASE2.is_file() else 'phase1'
    archive_path = export_dir / f'{OUT.name}_{stage}.zip'
    temporary = archive_path.with_suffix('.zip.tmp')
    include = [OUT, MANIFEST, CONFIG, PROJECT / 'source_revision.txt',
               PROJECT / 'qjepa', PROJECT / 'configs', PROJECT / 'tools',
               PROJECT / 'pyproject.toml', PROJECT / 'README.md']
    seen = set()
    with zipfile.ZipFile(temporary, 'w', compression=zipfile.ZIP_DEFLATED, compresslevel=1) as bundle:
        for path in include:
            for file in sorted(path.rglob('*')) if path.is_dir() else [path]:
                if not file.is_file() or '__pycache__' in file.parts or file.suffix == '.tmp':
                    continue
                name = file.relative_to(PROJECT).as_posix()
                if name not in seen:
                    bundle.write(file, name)
                    seen.add(name)
    temporary.replace(archive_path)
    print('Archive:', archive_path, '| GiB:', round(archive_path.stat().st_size / 2**30, 3))
    print('Tải file này hoặc lưu Kaggle Output trước khi kết thúc phiên.')
    display(FileLink(archive_path.name, url_prefix='/kaggle/working/'))
    return archive_path
```

## Cell 8 — Train phase 1

Đã có checkpoint phù hợp thì resume; đã đủ update thì bỏ qua train. Hash lệch thì
dừng, không xóa checkpoint. Run chết trước checkpoint đầu được đổi tên để giữ log.
Notebook dùng 3 worker (`DATALOADER_WORKERS` ở Cell 1) để tạo dữ liệu hỏng song song
với GPU; máy T4×2 có 4 vCPU. Theo dõi `RSS ... + worker ...` ở mỗi checkpoint. Nếu RAM
tiến gần giới hạn, interrupt sau checkpoint, hạ `DATALOADER_WORKERS` xuống 2, chạy lại
Cell 4 và resume. Worker/pin memory nằm ngoài configuration hash nên không mất checkpoint.

**Tự khởi động lại khi RAM cao.** Tiến trình train phase 2 của p11 bị kill ở update 4097 vì
RAM tăng ~6,7 MiB mỗi update (rò ở đường chạy GPU). Từ giờ, qua `runtime.restart_above_rss_gib`
(16 GiB), CLI lưu checkpoint rồi tự thoát với mã 75 và `train_phase` tự resume — kết quả giống
hệt một lần chạy liền (đã kiểm tra: trọng số trùng từng bit). Nếu tiến trình vẫn bị kill (-9)
nhưng đã có checkpoint mới hơn, `train_phase` cũng tự chạy tiếp; không có tiến triển thì dừng.


```python
RESTART_EXIT_CODE = 75   # CLI: da luu checkpoint, thoat vi RSS vuot runtime.restart_above_rss_gib
MAX_RESTARTS = 20

def checkpoint_update(checkpoint, phase):
    """So update cua checkpoint (0 neu chua co), sau khi kiem tra no thuoc dung run nay."""
    if not checkpoint.exists():
        return 0
    payload = load_checkpoint(checkpoint, 'cpu')
    metadata = dict(payload['metadata'])
    update = int(payload['successful_updates'])
    del payload
    gc.collect()
    assert metadata['configuration_hash'] == configuration_hash(config, phase), (
        f'{phase}: checkpoint khác config; chọn run mới, không xóa checkpoint này.'
    )
    assert metadata['manifest_hash'] == MANIFEST_HASH, 'Manifest khác checkpoint.'
    if phase == 'phase2':
        assert metadata['frozen_backbone_hash'] == PHASE1_BACKBONE_HASH, 'Phase 2 thuộc backbone khác.'
    return update

def train_phase(phase):
    if (config['runtime'].get('gpu_count'), config['runtime'].get('parallel')) != (GPU_COUNT, PARALLEL):
        # Phase truoc lui ve 1 GPU (xem ben duoi): phase nay van thu lai so GPU cua Cell 1.
        # runtime nam ngoai hash, nen checkpoint va resume khong doi.
        print(f'{phase}: dat lai GPU_COUNT={GPU_COUNT}, PARALLEL={PARALLEL}', flush=True)
        config['runtime'].update(gpu_count=GPU_COUNT, parallel=PARALLEL)
        encoded = yaml.safe_dump(config, sort_keys=False)
        CONFIG.write_text(encoded, encoding='utf-8')
        (OUT / 'config.yaml').write_text(encoded, encoding='utf-8')
    current = yaml.safe_load(CONFIG.read_text())
    for name in ('phase1', 'phase2'):
        assert configuration_hash(current, name) == configuration_hash(config, name), (
            'Config trên đĩa khác biến notebook; chạy lại Cell 4.'
        )
    folder = OUT / phase
    checkpoint = folder / 'last.pt'
    maximum = config[phase]['max_successful_updates']
    started = time.perf_counter()
    for attempt in range(MAX_RESTARTS + 1):
        update = checkpoint_update(checkpoint, phase)
        assert update <= maximum, 'Checkpoint có nhiều update hơn config.'
        if update == maximum:
            print(f'{phase}: đã hoàn tất {update}/{maximum}' + (', bỏ qua train.' if attempt == 0 else '.'))
            break
        if checkpoint.exists():
            print(f'{phase}: resume từ {update}/{maximum}')
        elif (folder / 'train.jsonl').exists():
            archived = folder.with_name(phase + '_incomplete_' + str(time.time_ns()))
            folder.rename(archived)
            print('Chưa có checkpoint; giữ log run dở tại', archived)
        arguments = ['train-' + phase, '--config', CONFIG, '--manifest', MANIFEST, '--output', OUT]
        if phase == 'phase2':
            arguments += ['--backbone-checkpoint', PHASE1]
        if checkpoint.exists():
            arguments += ['--resume', checkpoint]
        try:
            run_cli(*arguments)
            break
        except subprocess.CalledProcessError as error:
            # 75: CLI tu thoat SAU khi luu checkpoint vi RAM cao (ro bo nho o duong GPU,
            # ~6,7 MiB/update o p11). -9: bi kernel kill vi het RAM — chi chay lai khi da
            # co checkpoint moi hon, neu khong se lap mai o cung mot cho.
            progressed = checkpoint_update(checkpoint, phase) > update
            if error.returncode == RESTART_EXIT_CODE or (error.returncode == -9 and progressed):
                print(f'{phase}: khởi động lại tiến trình train để giải phóng RAM '
                      f'(lần {attempt + 1}/{MAX_RESTARTS}, mã {error.returncode})', flush=True)
                gc.collect()
                continue
            if config['runtime'].get('parallel') == 'ddp' and not progressed:
                # Ca phase 1 va phase 2 chay DDP khi GPU_COUNT = '2'. Chet truoc checkpoint moi thi
                # chay tiep tren 1 GPU. runtime nam ngoai hash: cung run, resume dung cho cu.
                print(f'{phase}: DDP 2 GPU loi (ma {error.returncode}) -> chay lai tren 1 GPU', flush=True)
                config['runtime'].update(gpu_count='1', parallel='data_parallel')
                encoded = yaml.safe_dump(config, sort_keys=False)
                CONFIG.write_text(encoded, encoding='utf-8')
                (OUT / 'config.yaml').write_text(encoded, encoding='utf-8')
                gc.collect()
                continue
            raise
    else:
        raise RuntimeError(f'{phase}: vượt {MAX_RESTARTS} lần khởi động lại; xem RSS trong train.jsonl.')
    print(phase, 'phút:', round((time.perf_counter() - started) / 60, 1))

train_phase('phase1')
```

## Cell 16 — Xem một đoạn liên tục: ảnh + IMU, với nhiễu y như lúc train

Mỗi lần chạy chọn **một đoạn liên tục ngẫu nhiên** dài `SECONDS` (mặc định 4 s) trong một trajectory
ngẫu nhiên của `SPLIT` (mặc định `valid`, ảnh model chưa thấy), rồi **bốc nhiễu mới đúng như lúc train**
(`fixed_realization=False`: cùng bộ làm hỏng ảnh và IMU, cùng xác suất, kể cả 2% frame để sạch và biến
thể "chỉ nhiễu" / "chỉ tối" của nhiễu B). Model đọc từng frame trong đoạn: hiện `PANELS` ảnh rải đều
(sạch · nhiễu · khôi phục, kèm PSNR và "vật nhỏ 2–4 px đúng chỗ"), và IMU của cả đoạn — các cửa sổ
chồng nhau được ghép như Cell 15.

Không cần chạy Cell 15 trước; chỉ cần Cell 1 → 4 và checkpoint `BEST`. Model được nạp một lần cho cả
phiên. Seed được in ra: `show_restoration(seed)` xem lại đúng đoạn và đúng lần bốc nhiễu đó.

Chú thích mỗi trục IMU có thêm **tỉ lệ rung** = sai số giữa hai mẫu liền kề chia cho sai số tổng. Gần
`1.0` nghĩa là lỗi là **rung**, không phải lệch.

```python
import secrets
import torch
from qjepa.cli import _dataset, _system_from_phase2
from qjepa.data import collate_paired, read_manifest
from qjepa.evaluation import ImuOverlapMerger, image_metrics

SECONDS = 4.0          # do dai DOAN LIEN TUC duoc xem (anh + IMU). Tang de nhin xu huong.
PANELS = 6             # so anh hien thi, rai deu trong doan
SPLIT = 'valid'        # anh model chua thay luc train ('test' cung duoc)
SEED = None            # None = moi lan chay: doan khac + lan boc nhieu khac; dat so de xem lai

# Nap model mot lan cho ca phien; chay lai cell khong nap lai tru khi BEST doi.
if globals().get('_preview_key') != str(BEST):
    _preview_device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    _preview_system, _preview_config = _system_from_phase2(str(BEST), _preview_device)
    # fixed_realization=False = DUNG nhieu luc train: cung bo lam hong anh + IMU, cung xac
    # suat (ca 2% frame de sach), cung bien the "chi nhieu" / "chi toi" cua nhieu B.
    _preview_dataset = _dataset(_preview_config, read_manifest(MANIFEST), SPLIT, fixed_realization=False)
    _preview_key = str(BEST)


def _variant(params):
    if params.get('clean'):
        return 'frame SACH (2% luc train)'
    if not params.get('low_light', True):
        return 'CHI NHIEU, khong toi'
    if not params.get('sensor_noise', True):
        return 'CHI TOI, khong hat'
    return 'toi + nhieu'


def show_restoration(seed=None):
    seed = secrets.randbelow(1_000_000) if seed is None else int(seed)
    rng = np.random.default_rng(seed)
    dataset = _preview_dataset
    # Moi lan chay mot lan boc nhieu moi, nhu luc train doi realization moi epoch.
    dataset.set_realization(int(rng.integers(1, 2**31 - 1)))
    print(f'seed {seed}  (chay lai dung doan + dung nhieu nay: show_restoration({seed}))')

    # Doan ngau nhien dai SECONDS trong mot trajectory ngau nhien.
    by_trajectory = {}
    for index, sample in enumerate(dataset.samples):
        by_trajectory.setdefault(sample.trajectory_key, []).append(index)
    traj = sorted(by_trajectory)[int(rng.integers(len(by_trajectory)))]
    frames = sorted(by_trajectory[traj], key=lambda i: dataset.samples[i].image_time)
    stamps = np.array([dataset.samples[i].image_time for i in frames])
    begin = stamps[0] + rng.uniform(0.0, max(0.0, stamps[-1] - stamps[0] - SECONDS))
    inside = [i for i, t in zip(frames, stamps) if begin <= t <= begin + SECONDS] or frames[:2]

    # Khoi phuc tung frame; IMU cua cac cua so chong nhau duoc ghep nhu Cell 15.
    _, imu_times_all = dataset.cache.get(dataset.samples[inside[0]])
    mergers = {name: ImuOverlapMerger(len(imu_times_all)) for name in ('clean', 'noisy', 'restored')}
    wanted = set(np.linspace(0, len(inside) - 1, min(PANELS, len(inside))).round().astype(int).tolist())
    panels = []
    for first in range(0, len(inside), 8):
        items = [dataset[i] for i in inside[first:first + 8]]
        batch = collate_paired(items)
        with torch.no_grad():
            out = _preview_system(*(batch[key].to(_preview_device) for key in
                                    ('image_noisy', 'imu_noisy_phys', 'image_time', 'imu_times')))
        for row, item in enumerate(items):
            start = int(item['imu_start'])
            mergers['clean'].add(start, batch['imu_clean_phys'][row].T.numpy())
            mergers['noisy'].add(start, batch['imu_noisy_phys'][row].T.numpy())
            mergers['restored'].add(start, out.imu_physical[row].T.cpu().numpy())
            if first + row in wanted:
                restored_image = out.image[row:row + 1].clamp(0, 1).cpu()
                panels.append((item['sample_id'], batch['image_clean'][row:row + 1],
                               batch['image_noisy'][row:row + 1], restored_image, item['corruption']['image']))

    print(f'\n{traj} | doan {SECONDS:g}s tu t={begin - stamps[0]:.1f}s | {len(inside)} frame'
          f' | {len(panels)} anh hien thi (nhieu moi, giong luc train)')
    fig, axes = plt.subplots(len(panels), 3, figsize=(12, 4.1 * len(panels)), squeeze=False,
                             facecolor='#fcfcfb', constrained_layout=True)
    for axis_row, (name, clean_image, noisy_image, restored_image, params) in zip(axes, panels):
        before, after = image_metrics(noisy_image, clean_image), image_metrics(restored_image, clean_image)
        titles = ('Sach (dich)', f'Nhieu luc train — {_variant(params)}', 'Khoi phuc')
        for axis, image, title in zip(axis_row, (clean_image, noisy_image, restored_image), titles):
            axis.imshow(image[0].permute(1, 2, 0).clamp(0, 1).numpy())
            axis.set_title(title, fontsize=9.5, loc='left')
            axis.set_xticks([]); axis.set_yticks([])
        axis_row[0].set_ylabel(name, fontsize=8)
        axis_row[2].set_xlabel(f"PSNR {before['image_psnr_db']:.2f} -> {after['image_psnr_db']:.2f} dB · "
                               f"vat nho 2-4px dung cho {before['image_fine_detail_in_place']:.2f} -> "
                               f"{after['image_fine_detail_in_place']:.2f}", fontsize=8.5)
    plt.show()
    plt.close(fig)

    # IMU cua cung doan do.
    CHANNELS = ('ax', 'ay', 'az', 'gx', 'gy', 'gz')
    SURFACE, INK, INK2 = '#fcfcfb', '#0b0b0b', '#52514e'
    CLEAN, NOISY, RESTORED = '#52514e', '#eb6834', '#2a78d6'
    clean, clean_cov = mergers['clean'].result()
    noisy, noisy_cov = mergers['noisy'].result()
    restored, restored_cov = mergers['restored'].result()
    coverage = clean_cov & noisy_cov & restored_cov
    w = coverage & (imu_times_all >= begin) & (imu_times_all <= begin + SECONDS)
    if w.sum() < 10:
        w = coverage
    t = imu_times_all[w] - imu_times_all[w][0]
    print(f'IMU: {int(w.sum())} mau trong {t[-1]:.1f}s')

    fig, axes = plt.subplots(6, 1, figsize=(13, 11), sharex=True,
                             constrained_layout=True, facecolor=SURFACE)
    for ch, ax in enumerate(axes):
        ax.set_facecolor(SURFACE)
        ax.plot(t, noisy[w, ch],    color=NOISY,    lw=1.2, alpha=0.75, label='Nhieu (input, nhu luc train)')
        ax.plot(t, clean[w, ch],    color=CLEAN,    lw=2.0, label='Sach (dich)')
        ax.plot(t, restored[w, ch], color=RESTORED, lw=2.0, label='Khoi phuc')

        e_in  = np.sqrt(np.mean((noisy[w, ch]    - clean[w, ch]) ** 2))
        e_out = np.sqrt(np.mean((restored[w, ch] - clean[w, ch]) ** 2))
        # Sai so giua hai mau lien ke: ~1.0 lan sai so tong nghia la RUNG chu
        # khong phai lech, va khong so hang loss theo tung mau nao nhin thay no.
        d_in  = np.sqrt(np.mean(np.diff(noisy[w, ch]    - clean[w, ch]) ** 2))
        d_out = np.sqrt(np.mean(np.diff(restored[w, ch] - clean[w, ch]) ** 2))
        ax.annotate(f"RMSE  input {e_in:.3f}  ->  khoi phuc {e_out:.3f}"
                    f"   {'v' if e_out < e_in else '^'} {abs(1 - e_out / max(e_in, 1e-9)) * 100:.0f}%"
                    f"      rung {d_in / max(e_in, 1e-9):.2f} -> {d_out / max(e_out, 1e-9):.2f}",
                    xy=(0.995, 0.06), xycoords='axes fraction', ha='right', va='bottom',
                    fontsize=8.5, color=INK2)
        ax.set_ylabel(f"{CHANNELS[ch]}\n" + ('m/s2' if ch < 3 else 'rad/s'), fontsize=9, color=INK)
        ax.grid(alpha=0.18, linewidth=0.6)
        for s in ('top', 'right'):
            ax.spines[s].set_visible(False)
        for s in ('left', 'bottom'):
            ax.spines[s].set_color('#d8d7d2')
        ax.tick_params(colors=INK2, labelsize=8)

    axes[-1].set_xlabel('Thoi gian trong doan (s)', fontsize=9, color=INK)
    axes[0].legend(loc='upper left', bbox_to_anchor=(0, 1.55), ncol=3,
                   frameon=False, fontsize=9.5, labelcolor=INK)
    fig.suptitle(f'{SPLIT} · nhieu luc train · IMU doan {SECONDS:g}s · {traj} · seed {seed}',
                 fontsize=11.5, color=INK, y=1.03)
    plt.show()
    plt.close(fig)

show_restoration(SEED)
gc.collect()
```

```python
import subprocess
import sys
import urllib.request
from pathlib import Path

# Lấy từ Cell 1/Cell 4 nếu có; nếu chạy session mới thì sửa hai dòng này.
PROJECT = Path(globals().get('PROJECT', '/kaggle/working/qwt_jepa_version3'))
RUN = Path(globals().get('OUT', PROJECT / 'outputs/p19_sharp_p2'))

# Checkout có thể đang ghim ở commit cũ chưa có tool; lúc đó tải bản mới nhất về.
script = PROJECT / 'tools/training_report.py'
if not script.is_file():
    script = Path('/kaggle/working/training_report.py')
    urllib.request.urlretrieve(
        'https://raw.githubusercontent.com/leesanghyoek/qwt_jepa_ver5/main/tools/training_report.py',
        script)
    print('Checkout chưa có tool; đã tải bản mới nhất về', script)

# RUN phải là thư mục CHỨA phase1/ và phase2/, không phải chính phase1/.
if not any((RUN / f'phase{n}/train.jsonl').is_file() for n in (1, 2)):
    print(f'Không thấy train.jsonl dưới {RUN}. Các run tìm được:')
    found = sorted({p.parent.parent for root in ('/kaggle/working', '/kaggle/input')
                    for p in Path(root).glob('**/phase*/train.jsonl')})
    for candidate in found:
        phases = sorted(q.parent.name for q in candidate.glob('phase*/train.jsonl'))
        print(f'   {candidate}   ({", ".join(phases)})')
    raise SystemExit('Đặt RUN = <một trong các đường dẫn trên> rồi chạy lại cell này.')

result = subprocess.run([sys.executable, '-u', str(script), '--run', str(RUN)],
                        capture_output=True, text=True)
print(result.stdout)
if result.stderr.strip():
    print('--- stderr ---')
    print(result.stderr)

report = Path('/kaggle/working/training_report.txt')
report.write_text(result.stdout, encoding='utf-8')
print(f'Đã lưu {report}  ({len(result.stdout)} ký tự) — tải về hoặc copy phần in ở trên.')
try:
    from IPython.display import FileLink, display
    display(FileLink(str(report)))
except Exception:
    pass
```
